# Anomaly Detection Prototyping

Validating detectors against the SAML-D and PaySim samples to ensure normalized score ranges span [0,1] without clustering.

```python
import pandas as pd
import matplotlib.pyplot as plt
import sys
import os
sys.path.append('..')

from agent.nodes.anomaly_node import run_anomaly_detection

# Load samples (100k rows for speed)
df_saml = pd.read_csv('../data/SAML-D.csv', nrows=100000)
df_paysim = pd.read_csv('../data/PS_20174392719_1491204439457_log.csv', nrows=100000)

print("SAML-D Sample Shape:", df_saml.shape)
print("PaySim Sample Shape:", df_paysim.shape)
```

```python
def plot_scores(res, title):
    scores = list(res['anomaly_scores'].values())
    plt.hist(scores, bins=50, color='skyblue', edgecolor='black')
    plt.title(title)
    plt.xlabel('Normalized Score')
    plt.ylabel('Frequency')
    plt.xlim(0, 1)
    plt.show()
    
    print(f"Method: {res['method_used']}")
    print(f"Flagged Transactions: {len(res['flagged_transactions'])}")
    if res['flagged_transactions']:
        print("Sample Flagged:", res['flagged_transactions'][0])
```

### 1. Test: Structuring Pattern (SAML-D)

```python
res_structuring = run_anomaly_detection(df_saml, 'structuring')
plot_scores(res_structuring, 'Structuring Pattern Scores (SAML-D)')
```

### 2. Test: Layering Pattern (SAML-D)

```python
res_layering = run_anomaly_detection(df_saml, 'layering')
plot_scores(res_layering, 'Layering Pattern Scores (SAML-D)')
```

### 3. Test: Cash Out Pattern (PaySim)

```python
res_cash_out = run_anomaly_detection(df_paysim, 'cash_out')
plot_scores(res_cash_out, 'Cash Out Pattern Scores (PaySim)')
```

### 4. Test: Broad Exploration ('none' Pattern on PaySim)

```python
res_none = run_anomaly_detection(df_paysim, 'none')
plot_scores(res_none, 'Broad Exploration Scores (PaySim)')
```
